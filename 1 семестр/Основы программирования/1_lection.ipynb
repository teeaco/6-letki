<h1><b>Программирование, как процесс процедурного решения задач</b></h1>Считаю, что корректнее для начала разобраться, что такое программирование в целом<br/>------<br/><b>Программирование - это процесс создания компьтерных программ для решения каких-либо задач</b><br/>------<br/>Мы прекрасно понимаем, что задачи у нас могут быть разнообразные, начиная от ускорения какого-то процесса заканчивая созданием какого-то продукта (программы), который закроет новые потребности человечества. Употребила термин "программа", поэтому даю определение<br/>------<br/><b>Программа — это набор инструкций, предписаний или планов, определяющих последовательность действий для достижения определённой цели</b><br/>------

Программисты - люди умные и давным-давно поняли, что для разных поставленных задач, нужно использовать разные подходы (очевидно?).<br/>Их давно придумали, разработали и классифицировали. А обозвали эти подходы красивым названием "Парадигмы"<br/>------<br/><b>Парадигмы - набор идей, концепций и правил, определяющих стиль написания компьютерных программ и способ организации вычислений</b><br/>------<br/>У каждой парадигмы есть свои правила, стандарты и понятия, которым обязан следовать разработчик. Хочу также отметить, что одна и та же задача может быть решена разными парадигмами, так что периодически перед вам будет стоять задача "С помощью какой парадигмы мне решить эту задачу?"

<b>Классификация парадигм</b><br>
Сейчас будет длинный текст, разъясняющий что для чего, потом увидите красивую картинку, благодаря которой вам станет все понятно<br>Две основные парадигмы: импертивная и декларативная, которые вмещают в себя остальные (то есть остальные парадигмы - подтипы этих двух)<br><br>= Императивная<br>Самая простая и часто используемая парадигма. Её смысл в последовательном выполнении действий. Её основные подтипы, которые для вас будут иметь интерес: Процедурное и Объектно-ориентированное программирование (ООП)<br>- Процедурное<br>Основным понятие в процедурном программирования является функция (процедура)<br>------<br><b>Функция (или процедура) - последовательность действий, которую записали и назвали</b><br>------<br>В этой парадигме вы будете работать в этом семестре<br>- ООП<br>В этой парадигме есть такое понятие, как объект, который обладает сам какими-то свойствами и функциями. Об этом подробнее в следующем семестре.<br>P.S. Пример процедурного: вам прислали шкаф с инструкцией по сборке. В данной ситуации, инструкция к шкафу - есть функция (процедура)<br>Если говорить про ооп, то в этом подходе у нас есть объект шкаф, который можно собрать, разобрать и т.д. В этой парадигме центром программы будут объекты и их взаимодействие, а не сами функции<br><br>= Декларативная<br>В этой парадигме разработчику сосредатачивается на результате, а не на самих действиях. Т.е. разработчик дает команду, а компьютер ищет сам решение для этой задачи. Если в императивной разработчик обязательно самостоятельно указывает последовательность действий, то в декларативной система сама решит как это выполнить.<br>- Функциональное программирование<br>НЕ ПУТАТЬ С ПРОЦЕДУРНЫМ, ЭТО РАЗНЫЕ ПАРАДИГМЫ. В строгом функциональном подходе считается, что все функции чистые и не имеют побочных эффектов. Функция никогда не меняет объект, а возвращает всегда новый<br>- Логическое программирование<br>Это такой подход к созданию программ, основанный на принципах математической логики, где программа представляет собой набор фактов и правил вывода, а ее выполнение заключается в автоматическом выводе заключений на основе этих фактов<br><br>В заключении по этому блоку хочу отметить, что есть концепции и принципы, которые пересекаются в языках. <h1><b>Понятие язык программирования</b></h1>------<br><b>Язык программирования (ЯП) — это формальная знаковая система, на которой пишут компьютерные программы</b><br>------<br>Для компьютера существует один единственный понятный ему код - двоичный. Но писать на нем не очень удобно, не находите? Для этого и были созданы языки программирования. Классифицировать ЯП можно двумя способами:<br><br><i>Компилируемые и Интерпертируемые ЯП</i><br>Они различаются принципом работы.
Компилятор получает на вход код и компилирует его — собирает, преобразует в исполняемый файл весь разом. Потом этот исполняемый файл можно запустить, и компьютер поймет, что от него хотят.
Интерпретатор «переводит» код построчно и тут же инициирует его выполнение. Сначала он прочитает и отправит на исполнение первую строчку, затем вторую – и так далее. Такая классификация важна для отладки кода.<br>Немного подробнее про этапы компиляции, т.к. процесс это сложный, но безумно важный для понимания<br>
1. <i>Препроцессирование.</i> Вставление директив, статических библиотек и модулей<br>
2. <i>Трансляция.</i> «Трансляция» — это (с английского) «перевод». Кого и куда переводим? Наш язык в ассемблер. (это язык, самый близкий к машинному коду)<br>
3. <i>Ассемблирование.</i> На данном этапе не происходит ничего интересного — просто инструкции, которые были в ассемблере, перегоняются в машинный код.<br>
4. <i>Линковка.</i> На выходе даёт исполняемый файл. Нужна, если файлов несколько: мы запускаем препроцессор, трансляцию и ассемблирование независимо для каждого файла, а объединяются они только на этапе линковки.<br>
5. <i>Загрузка.</i> Последний этап, который предстоит пройти нашей программе — вызвать загрузчик для загрузки нашей программы в память. На данной стадии также возможна подгрузка динамических библиотек.<br><br><i>Низкоуровневые и Высокоуровневые ЯП</i><br>У низкоуровневых языков нет встроенных функций. Они общаются с конкретным процессором, поэтому код одного компьютера не может быть использован на другом. У каждого производителя процессоров свой собственный набор инструкций. 
Самые распространенные низкоуровневые языки — это машинный код и язык ассемблера.<br>Язык ассемблера упрощает машинный код, но он все еще сложен для массового использования. Поэтому программисты начали создавать новые языки, которые больше упрощают программирование. Так появились высокоуровневые языки, синтаксис которых близок к английскому.<br>Важно понимать нюанс- чем ниже язык по уровню, тем больше свободы и меньше затрат по времени и памяти языку необходимо. Например, одна и та же программа на ассемблере будет весить 16 кб, на С++ 4 мб, а на питоне уже 16 мб. Складывается впечатление, что питон нужно удалить из жизни всех людей по такой логике. Однако я не зря сказала про свободу. На ассемблере вы можете удалить папку с ОС, а на питоне это невозможно. На плюсах у вас больше возможностей по управлению памятью, а на том же питоне у вас чисто визически это не получится. Сильное преимущество таких языков как С++ - возможность выделять память, управлять ею, а также удалять. Для некоторых задач это крайне важно, поэтому мы вас и знакомим с таким чудесным языком с большим количеством возможностей<br>
<br>!На самом деле границу по высокому\низкому уровню установить практически невозможно, банально в виду того, что только машинный код компьютер понимает, поэтому много значения этой классификации не придавайте.

<h1><b>Структуры программы на языке программирования на примере языка С++</b></h1>Уже не терпится увидеть С++, но акцентирую внимание на следующем моменте: общей структуры для всех ЯП не существует, да, есть какие-то общие правила и тенденции, но вы их заметите при знакомстве с другими языками, поэтому говорить за все ЯП не буду<br>

```python
#include <iostream>               // подключаем заголовочный файл iostream
  
int main()                          // определяем функцию main
{                                   // начало функции
    std::cout << "Hello World!";   // выводим строку на консоль
    return 0;                       // выходим из функции
}                                   // конец функции
```

Это тот самый "Hello World!", который пишут все в самом начале. На нем разберемся с структурой программы на С++

Инструкции<br>
Программа на С++ состоит из набора инструкций. Каждая инструкция (statement) выполняет определенное действие. В конце инструкции в языке C++ ставится точка с запятой (;). Данный знак указывает компилятору на завершение инструкции. Например:

```python
std::cout << "Hello World!"; //является инструкцией и поэтому завершается точкой с запятой.
```

```python
{
    std::cout << "Hello World!"; //чтобы определить блок кода мы его заключаем в фигурные скобки
    std::cout << "Bye World!";   //а инструкции распологаем между ними
}   
```

Функция main<br>
Каждая программа на языке С++ должна иметь как минимум одну функцию - функцию main(). Именно с этой функции начинается выполнение приложения. Имя main фиксировано и для всех программ на С++ всегда одинаково.<br>
Функция также является блоком кода, поэтому ее тело обрамляется фигурными скобками, между которыми определяется набор инструкций. int main() мы как раз видим в нашем примере<br>
Определение функии main начинается с возвращаемого типа. Функция main в любом случае должна возвращать число, поэтому тип int. 
Далее идет название функции, то есть main. После названия в скобках идет список параметров. В данном случае функция main не принимает никаких параметров, поэтому после названия указаны пустые скобки.<br>
Далее идут как раз наши действия<br>
В конце функции идет инструкция return, обозначающая завершение выполнения функции и вовзращающая какое-то значение в месте, где была вызвана<br>main запускает всегда, это, буквально, тело нашей программы, которая возвращает значение, внимание, самой ОС<br>
Число 0 после оператора return указывает операционной системе, что выполнение функции завершилось успешно, без ошибок.

Директивы препроцессора<br>
В примере выше на консоль выводится строка, но чтобы использовать вывод на консоль, необходимо в начале файла с исходным кодом подключать библиотеку iostream с помощью директивы include.<br>
В отличие от обычных инструкциий языка C++, которые завершаются точкой с запятой ; , признаком завершения препроцессорной директивы является перевод на новую строку. Кроме того, директива должна начинаться со знака решетки #. Непосредственно директива "include" определяет, какие файлы и библиотеки надо подключить в данном месте в код программы.
<br>
Комментарии<br><br>
Комментарии нужны как чужим людям, смотрящие ваш код, так и вам самому, чтобы понять, что вы написали в 5 утра<br>Комментарии игнорируются компилятором и не влияют на решение, вес и время выполнения кода.<br>
В языке C++ есть два типа комментариев: однострочный и многострочный. Однострочный комментарий размещается на одной строке после двойного слеша // (его вы уже видели)<br>
Многострочный комментарий заключается между символами /* текст комментария */. Он может размещаться на нескольких строках. 

```python
#include <iostream>
/*
    Определение функции Main
    Выводит на консоль строку Hello World!
*/
int main()
{
    std::cout << "Hello World!"; // вывод строки на консоль
    return 0;
}
```
